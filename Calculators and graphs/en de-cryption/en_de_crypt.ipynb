1. AES Encryption/Decryption (Python using pycryptodome):

```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad

# Encryption
key = get_random_bytes(16)  # 128-bit key
cipher = AES.new(key, AES.MODE_CBC)
plaintext = b"This is a secret message"
ciphertext = cipher.encrypt(pad(plaintext, AES.block_size))
iv = cipher.iv  # Save the initialization vector

print("Ciphertext:", ciphertext)

# Decryption
cipher = AES.new(key, AES.MODE_CBC, iv)
decrypted = unpad(cipher.decrypt(ciphertext), AES.block_size)

print("Decrypted:", decrypted.decode())
```

2. RSA Encryption/Decryption (Python using pycryptodome):

```python
from Crypto.PublicKey import RSA
from Crypto.Cipher import PKCS1_OAEP

# Key generation
key = RSA.generate(2048)
public_key = key.publickey()

# Encryption
cipher_rsa = PKCS1_OAEP.new(public_key)
message = b"This is a secret message"
ciphertext = cipher_rsa.encrypt(message)

print("Ciphertext:", ciphertext)

# Decryption
cipher_rsa = PKCS1_OAEP.new(key)
decrypted = cipher_rsa.decrypt(ciphertext)

print("Decrypted:", decrypted.decode())
```

3. Blowfish Encryption/Decryption (Python using pycryptodome):

```python
from Crypto.Cipher import Blowfish
from Crypto.Util.Padding import pad, unpad

key = b'An arbitrarily long key'
cipher = Blowfish.new(key, Blowfish.MODE_CBC)
plaintext = b"Secret Message"
ciphertext = cipher.encrypt(pad(plaintext, Blowfish.block_size))
iv = cipher.iv

print("Ciphertext:", ciphertext)

cipher = Blowfish.new(key, Blowfish.MODE_CBC, iv)
decrypted = unpad(cipher.decrypt(ciphertext), Blowfish.block_size)

print("Decrypted:", decrypted.decode())
```

4. ChaCha20 Encryption/Decryption (Python using pycryptodome):

```python
from Crypto.Cipher import ChaCha20
from Crypto.Random import get_random_bytes

key = get_random_bytes(32)  # 256-bit key
cipher = ChaCha20.new(key=key)
plaintext = b"Secret message"
ciphertext = cipher.encrypt(plaintext)
nonce = cipher.nonce

print("Ciphertext:", ciphertext)

cipher = ChaCha20.new(key=key, nonce=nonce)
decrypted = cipher.decrypt(ciphertext)

print("Decrypted:", decrypted.decode())
```

Caesar Cipher (Python):

```python
def caesar_cipher(text, shift):
    result = ""
    for i in range(len(text)):
        char = text[i]
        if char.isupper():
            result += chr((ord(char) + shift - 65) % 26 + 65)
        else:
            result += chr((ord(char) + shift - 97) % 26 + 97)
    return result

text = "HELLO"
shift = 3
encrypted = caesar_cipher(text, shift)
print("Caesar Cipher:", encrypted)

# Decrypt by shifting in the opposite direction
decrypted = caesar_cipher(encrypted, -shift)
print("Decrypted:", decrypted)
```

Vigenère Cipher (Python):

```python
def vigenere_cipher(text, key):
    result = []
    key_length = len(key)
    for i, char in enumerate(text):
        shift = ord(key[i % key_length].upper()) - 65
        if char.isupper():
            result.append(chr((ord(char) + shift - 65) % 26 + 65))
        elif char.islower():
            result.append(chr((ord(char) + shift - 97) % 26 + 97))
        else:
            result.append(char)
    return "".join(result)

text = "HELLO"
key = "KEY"
encrypted = vigenere_cipher(text, key)
print("Vigenère Cipher:", encrypted)

# Decrypt by reversing the shift
decrypted = vigenere_cipher(encrypted, ''.join(chr((65 - (ord(k.upper()) - 65)) % 26 + 65) for k in key))
print("Decrypted:", decrypted)
```

Monoalphabetic Substitution Cipher (Python):

```python
import string

def monoalphabetic_cipher(text, key):
    alphabet = string.ascii_uppercase
    result = text.upper().translate(str.maketrans(alphabet, key))
    return result

text = "HELLO"
key = "QWERTYUIOPLKJHGFDSAZXCVBNM"  # Example substitution key
encrypted = monoalphabetic_cipher(text, key)
print("Monoalphabetic Cipher:", encrypted)

# Decrypt by reversing the key
decrypted = monoalphabetic_cipher(encrypted, ''.join(sorted(key, key=lambda k: key.index(k))))
print("Decrypted:", decrypted)
```

Playfair Cipher (Python):

```python
def generate_playfair_key_matrix(key):
    matrix = []
    key = "".join(sorted(set(key), key=lambda k: key.index(k))).replace("J", "I")
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    for char in key:
        if char not in matrix:
            matrix.append(char)
    for char in alphabet:
        if char not in matrix:
            matrix.append(char)
    return [matrix[i:i + 5] for i in range(0, len(matrix), 5)]

def playfair_cipher(text, key, encrypt=True):
    matrix = generate_playfair_key_matrix(key)
    text = text.replace("J", "I").upper().replace(" ", "")
    if len(text) % 2 != 0:
        text += "X"
    digraphs = [text[i:i + 2] for i in range(0, len(text), 2)]
    
    result = []
    for digraph in digraphs:
        pos1, pos2 = None, None
        for row in matrix:
            if digraph[0] in row:
                pos1 = (matrix.index(row), row.index(digraph[0]))
            if digraph[1] in row:
                pos2 = (matrix.index(row), row.index(digraph[1]))
        if pos1[0] == pos2[0]:
            result.append(matrix[pos1[0]][(pos1[1] + (1 if encrypt else -1)) % 5])
            result.append(matrix[pos2[0]][(pos2[1] + (1 if encrypt else -1)) % 5])
        elif pos1[1] == pos2[1]:
            result.append(matrix[(pos1[0] + (1 if encrypt else -1)) % 5][pos1[1]])
            result.append(matrix[(pos2[0] + (1 if encrypt else -1)) % 5][pos2[1]])
        else:
            result.append(matrix[pos1[0]][pos2[1]])
            result.append(matrix[pos2[0]][pos1[1]])
    return "".join(result)

text = "HELLO"
key = "PLAYFAIR"
encrypted = playfair_cipher(text, key)
print("Playfair Cipher:", encrypted)

# Decrypt the Playfair cipher
decrypted = playfair_cipher(encrypted, key, encrypt=False)
print("Decrypted:", decrypted)
```

DES (Data Encryption Standard) (Python using pycryptodome):

```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes

key = get_random_bytes(8)  # DES uses a 64-bit key (8 bytes)
cipher = DES.new(key, DES.MODE_CBC)
plaintext = b"Secret Message"
ciphertext = cipher.encrypt(pad(plaintext, DES.block_size))
iv = cipher.iv

print("Ciphertext:", ciphertext)

cipher = DES.new(key, DES.MODE_CBC, iv)
decrypted = unpad(cipher.decrypt(ciphertext), DES.block_size)

print("Decrypted:", decrypted.decode())
```

```python
Practical Implementation of Diffie-Hellman in Python
Here’s how you might implement Diffie-Hellman key exchange in Python using the cryptography library:
```

```python
from cryptography.hazmat.primitives.asymmetric import dh
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.backends import default_backend

# Parameters (publicly agreed upon)
parameters = dh.generate_parameters(generator=2, key_size=2048, backend=default_backend())

# Alice generates her private and public keys
alice_private_key = parameters.generate_private_key()
alice_public_key = alice_private_key.public_key()

# Bob generates his private and public keys
bob_private_key = parameters.generate_private_key()
bob_public_key = bob_private_key.public_key()

# Alice and Bob exchange their public keys and generate the shared secret
alice_shared_key = alice_private_key.exchange(bob_public_key)
bob_shared_key = bob_private_key.exchange(alice_public_key)

# The shared key should be the same for both Alice and Bob
assert alice_shared_key == bob_shared_key
print("Shared Key:", alice_shared_key.hex())

# Example: Deriving a key from the shared secret (e.g., for AES encryption)
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.kdf.hkdf import HKDF

derived_key = HKDF(
    algorithm=hashes.SHA256(),
    length=32,
    salt=None,
    info=b'handshake data',
    backend=default_backend()
).derive(alice_shared_key)

print("Derived AES Key:", derived_key.hex())
```
