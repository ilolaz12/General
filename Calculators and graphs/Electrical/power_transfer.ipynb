```python
import numpy as np
import matplotlib.pyplot as plt
Vs = 10
Rsrc = 100
Rload = np.arange (0, 1002 , 2)
P = (Vs**2)/(( Rsrc + Rload)**2)*Rload
[fig , ax] = plt. subplots ()
ax.plot(Rload , P)
ax.set(xlabel= '$R_{load}\, \Omega$', ylabel = 'Power W' , title= 'Power Transfer ' )
ax.grid(which= 'major', visible =True, linestyle = 'solid')
ax.grid(which= 'minor', visible =True, linestyle = 'dashed')
ax. minorticks_on ()
plt.show(block=False)
```

```python
import csv
import numpy as np
import matplotlib.pyplot as plt
FileName = ' FibreData_9 .txt '
with open(FileName , ' r ' ) as fh:
pwrlist = []
for row in csv.reader(fh , delimiter = ' , ' ):
pwrlist .append(row)
fh.close ()
pwr = np.array(pwrlist , dtype=float)
plt.plot(pwr)
plt.show(block=False)
plt. savefig ( ' OTDRresults .png ' )
```
