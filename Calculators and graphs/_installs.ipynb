Step 1: Install Jupyter Notebook and IPython Kernel
First, ensure you have Jupyter Notebook and the IPython kernel installed:


pip install notebook ipykernel

Step 2: Add Python Kernel to Jupyter
After installing Jupyter and IPython, add the Python kernel to Jupyter:

python -m ipykernel install --user

Step 3: Launch Jupyter Notebook
Start Jupyter Notebook:

jupyter notebook
