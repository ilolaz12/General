Finding derivatives in Python can be done using several libraries, depending on the complexity and type of function you're dealing with. Here's a guide for different levels of difficulty using SymPy for symbolic differentiation and NumPy for numerical differentiation.

Level 1: Basic Symbolic Differentiation using SymPy
SymPy is a Python library for symbolic mathematics. It's great for finding exact derivatives of functions.

Level 4: Numerical Differentiation using NumPy
For numerical differentiation, you can use NumPy. This is useful for functions defined by data points or those that are not easily differentiable symbolically.

Basic Derivative:

```python
import sympy as sp

# Define the symbol
x = sp.symbols('x')

# Define the function
f = x**2 + 3*x + 2

# Calculate the derivative
f_prime = sp.diff(f, x)

# Display the result
display(f_prime)
```

Level 2: Higher-Order Derivatives
For higher-order derivatives, you can use the diff function multiple times or specify the order directly.

```python
import sympy as sp

# Define the symbol
x = sp.symbols('x')

# Define the function
f = sp.sin(x) * sp.exp(x)

# Calculate the first derivative
f_prime = sp.diff(f, x)

# Calculate the second derivative
f_double_prime = sp.diff(f, x, 2)

# Display the results
display(f_prime)
display(f_double_prime)
```

Level 3: Partial Derivatives for Multivariable Functions
For functions with multiple variables, you can find partial derivatives.

```python
import sympy as sp

# Define the symbols
x, y = sp.symbols('x y')

# Define the function
f = x**2 * y + sp.sin(x*y)

# Calculate the partial derivatives
f_partial_x = sp.diff(f, x)
f_partial_y = sp.diff(f, y)

# Display the results
display(f_partial_x)
display(f_partial_y)
```

Level 4: Numerical Differentiation using NumPy
For numerical differentiation, you can use NumPy. This is useful for functions defined by data points or those that are not easily differentiable symbolically.

```python
import numpy as np

# Define the function
def f(x):
    return np.sin(x) * np.exp(x)

# Define the numerical derivative function
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

# Calculate the derivative at a point
x_val = 1.0
derivative_at_x = numerical_derivative(f, x_val)

# Display the result
print(f"The derivative of f at x = {x_val} is approximately {derivative_at_x}")
```

Level 5: Higher-Order Numerical Derivatives
To calculate higher-order numerical derivatives, you can extend the numerical differentiation method.

```python
import numpy as np

# Define the function
def f(x):
    return np.sin(x) * np.exp(x)

# Define the numerical derivative function
def numerical_derivative(f, x, h=1e-5, order=1):
    if order == 1:
        return (f(x + h) - f(x - h)) / (2 * h)
    elif order == 2:
        return (f(x + h) - 2 * f(x) + f(x - h)) / (h**2)
    else:
        raise ValueError("Higher-order derivatives not implemented")

# Calculate the first and second derivatives at a point
x_val = 1.0
first_derivative_at_x = numerical_derivative(f, x_val, order=1)
second_derivative_at_x = numerical_derivative(f, x_val, order=2)

# Display the results
print(f"The first derivative of f at x = {x_val} is approximately {first_derivative_at_x}")
print(f"The second derivative of f at x = {x_val} is approximately {second_derivative_at_x}")
```
