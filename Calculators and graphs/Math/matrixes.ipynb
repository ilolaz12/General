Basic Matrix Creation and Operations:

```python
import numpy as np

# Define matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Matrix addition
C = A + B
print("Matrix Addition:\n", C)

# Matrix multiplication
D = np.dot(A, B)
print("Matrix Multiplication:\n", D)

# Element-wise multiplication
E = A * B
print("Element-wise Multiplication:\n", E)

# Transpose of a matrix
F = np.transpose(A)
print("Transpose of Matrix A:\n", F)
```

Level 2: Determinant and Inverse of a Matrix

```python
import numpy as np

# Define matrix
A = np.array([[1, 2], [3, 4]])

# Determinant
det_A = np.linalg.det(A)
print("Determinant of A:", det_A)

# Inverse
inv_A = np.linalg.inv(A)
print("Inverse of A:\n", inv_A)
```

Level 3: Eigenvalues and Eigenvectors

```python
import numpy as np

# Define matrix
A = np.array([[1, 2], [3, 4]])

# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues of A:", eigenvalues)
print("Eigenvectors of A:\n", eigenvectors)
```

Level 4: Solving Linear Systems

```python
import numpy as np

# Define matrix A and vector b
A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])

# Solve the system of equations Ax = b
x = np.linalg.solve(A, b)
print("Solution x:", x)
```

Level 5: Advanced Matrix Operations
For more advanced operations such as matrix decompositions (QR, SVD, etc.), NumPy and SciPy provide robust functionality.

QR Decomposition

```python
import numpy as np

# Define matrix
A = np.array([[12, -51, 4], [6, 167, -68], [-4, 24, -41]])

# QR decomposition
Q, R = np.linalg.qr(A)
print("Q matrix:\n", Q)
print("R matrix:\n", R)
```

Singular Value Decomposition (SVD)

```python
import numpy as np

# Define matrix
A = np.array([[1, 2], [3, 4], [5, 6]])

# Singular Value Decomposition
U, S, Vt = np.linalg.svd(A)
print("U matrix:\n", U)
print("Singular values:", S)
print("Vt matrix:\n", Vt)
```

Sparse Matrices:

```python
import numpy as np
from scipy.sparse import csr_matrix

# Create a sparse matrix
A = csr_matrix([[0, 0, 3], [4, 0, 0], [0, 5, 6]])
print("Sparse Matrix A:\n", A)

# Convert sparse matrix to dense
A_dense = A.toarray()
print("Dense Matrix A:\n", A_dense)
```

Solving Sparse Linear Systems:

```python
import numpy as np
from scipy.sparse import csr_matrix
from scipy.sparse.linalg import spsolve

# Create a sparse matrix
A = csr_matrix([[3, 0, 0], [0, 4, 5], [0, 5, 6]])
b = np.array([9, 8, 7])

# Solve the system of equations Ax = b
x = spsolve(A, b)
print("Solution x:", x)
```
