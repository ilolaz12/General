The received signal at an antenna terminal is displayed as 9.1dBμV

If this voltage is developed across a 14Ω
load, calculate  the power expressed in dBm . 

```python
import math

# Given values
dBuV = 9.1
load_resistance_ohms = 14

# Convert voltage from dBμV to μV
voltage_microvolts = 10 ** (dBuV / 20)

# Convert voltage from μV to volts
voltage_volts = voltage_microvolts / 1e6

# Calculate power in watts using the formula P = V^2 / R
power_watts = (voltage_volts ** 2) / load_resistance_ohms

# Convert power from watts to milliwatts
power_milliwatts = power_watts * 1e3

# Convert power from mW to dBm
power_dbm = 10 * math.log10(power_milliwatts)

# Round the result to one decimal place
power_dbm_rounded = round(power_dbm, 1)
power_dbm_rounded
```

A radio channel has a bandwidth of 78MHz and a signal to noise ratio of 7dB
Calculate the theoretical capacity of the channel in Mbps. 

```python
# Given values
bandwidth_hz = 78e6  # Bandwidth in Hz
snr_db = 7  # Signal-to-noise ratio in dB

# Convert SNR from dB to linear
snr_linear = 10 ** (snr_db / 10)

# Apply the Shannon-Hartley theorem
channel_capacity_bps = bandwidth_hz * math.log2(1 + snr_linear)

# Convert channel capacity to Mbps
channel_capacity_mbps = channel_capacity_bps / 1e6
channel_capacity_mbps_rounded = round(channel_capacity_mbps, 1)
channel_capacity_mbps_rounded
```

A transmit-receive antenna pair are to be used for  short-range wireless coverage at a frequency of 3GHz. The transmit antenna is at a height of 2.7m, and the receive antenna is at a height of 1.1m. The distance between transmitter and receiver is 4.7m
Calculate the approximate path-length difference in m
between a direct and reflected radio wave propagation. 

```python
import math

# Given parameters
frequency_GHz = 3.0  # Frequency in GHz
height_transmit_m = 2.7  # Height of the transmit antenna in meters
height_receive_m = 1.1  # Height of the receive antenna in meters
distance_horizontal_m = 4.7  # Horizontal distance between the antennas in meters

# Calculate the direct path length using Pythagorean theorem
height_difference = abs(height_transmit_m - height_receive_m)
direct_path_length = math.sqrt(height_difference**2 + distance_horizontal_m**2)

# Calculate the reflected path length
# Assuming perfect reflection from the ground point midway between the antennas
reflected_path_length = 2 * math.sqrt((distance_horizontal_m / 2)**2 + (height_transmit_m + height_receive_m)**2)

# Calculate the path-length difference
path_length_difference = reflected_path_length - direct_path_length
round(path_length_difference, 2)  # Round the result to two decimal places

```

A multimode optical fibre has a core refractive index of 1.59 and a cladding refractive index of 1.42.  Suppose a ray just enters at an angle equal to the numerical aperture.  When this ray first reflects off the cladding, the angle of the ray as measured along the fibre axis would be calculated as:

```python
import math

nf = 1.59  # refractive index of fiber core
nc = 1.42  # refractive index of cladding

NA = math.sqrt(nf**2 - nc**2) #The Numerical Aperture (NA) of an optical fiber

# Calculate maximum non-reflective angle
max_non_ref_angle = math.asin(NA)

# Convert the angle from radians to degrees if needed
max_non_ref_angle_degrees = math.degrees(max_non_ref_angle)

print(max_non_ref_angle_degrees)
```

Explanation:
import: The from scapy.all import sniff, Packet statement imports the necessary functions from scapy.
process_packet: This function is called for each captured packet. It takes a packet as an argument and prints a summary of it.
capture_packets: This function sets up packet capturing. sniff is a function from scapy that captures packets. The count=10 argument specifies that it should stop after 10 packets. The prn=process_packet argument tells sniff to call process_packet for each packet it captures. store=False means it does not keep the packets in memory after processing (useful for long captures to save memory).
main block: Checks if the script is the main program and starts the capture

```python
from scapy.all import sniff, Packet

# Define the packet processing function
def process_packet(packet: Packet):
    print(packet.summary())

# Capture packets
def capture_packets():
    sniff(count=10, prn=process_packet, store=False)

if __name__ == "__main__":
    print("Starting packet capture...")
    capture_packets()
```
