PyMuPDF (fitz)
Overview:
PyMuPDF is a Python binding for MuPDF, which is a lightweight PDF, XPS, and E-book viewer. MuPDF can also render other document formats like EPUB, HTML, and SVG. PyMuPDF is known for its speed and efficiency in rendering pages and extracting content from documents.

Key Features:

PDF Manipulation: It supports operations like merging, splitting, rotating, and cropping PDF pages.
Text Extraction: Allows for high-quality extraction of text and formatting from PDFs.
Image Extraction and Insertion: You can extract images from documents or embed images into documents.
Annotation and Metadata: It supports reading and modifying annotations and PDF metadata.
Support for Multiple Formats: Apart from PDFs, it can handle XPS, OpenXPS, EPUB, and several other formats.
Usage Example:
Here is a simple example of using PyMuPDF to extract text from a PDF document:

```python
import fitz  # Import PyMuPDF

def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = ''
    for page in doc:
        text += page.get_text()
    doc.close()
    return text

pdf_text = extract_text_from_pdf('example.pdf')
print(pdf_text)
```

1. PDFMiner
Overview:
PDFMiner is a tool for extracting information from PDF documents. Unlike PyMuPDF, which is a general PDF tool, PDFMiner focuses more on extracting text and does not render pages.

Key Features:

Text Extraction with Layout: It can analyze the layout of a page and extract text according to the layout.
Conversion to Other Formats: It can convert PDF text into other formats like HTML and XML.
No Support for Writing PDFs: It’s purely for analysis and extraction, not for creating or editing PDFs.
Usage Example:
Extracting text with layout using PDFMiner:

```python
from pdfminer.high_level import extract_text

text = extract_text('example.pdf')
print(text)
```

2. PyPDF2
Overview:
PyPDF2 is a Pure-Python library built as a PDF toolkit. It is capable of splitting, merging, cropping, and transforming the pages of PDF files. It can also add custom data, viewing options, and passwords to PDFs.

Key Features:

Manipulate PDFs: Merge, split, and transform PDF documents.
Extract Information: Get information about the document, such as the number of pages, metadata, and text (although its text extraction capabilities are more limited compared to PyMuPDF or PDFMiner).
Usage Example:
Merging PDFs with PyPDF2:

```python
from PyPDF2 import PdfReader, PdfWriter

reader1 = PdfReader('file1.pdf')
reader2 = PdfReader('file2.pdf')
writer = PdfWriter()

for page in reader1.pages:
    writer.add_page(page)
for page in reader2.pages:
    writer.add_page(page)

with open('merged_file.pdf', 'wb') as f:
    writer.write(f)
```

These libraries offer different capabilities, and the choice between them would depend on your specific needs: PyMuPDF for general fast processing and rendering, PDFMiner for detailed text analysis and extraction, and PyPDF2 for simple PDF manipulations without needing the detailed text extraction capabilities

```python
import os
from openpyxl import load_workbook
import pandas as pd
import PyPDF2

def search_pdf(file_path, search_terms):
    results = {term: [] for term in search_terms}
    with open(file_path, "rb") as file:
        reader = PyPDF2.PdfReader(file)
        for i in range(len(reader.pages)):
            text = reader.pages[i].extract_text()
            if text:
                for term in search_terms:
                    if term in text:
                        results[term].append(i+1)
    return results

def export_to_excel(results, excel_path):
    data = [{'Term': term, 'Pages': ', '.join(map(str, pages))} for term, pages in results.items() if pages]
    df = pd.DataFrame(data)
    
    if not os.path.exists(excel_path):
        # If the file does not exist, create it and write directly
        with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
            df.to_excel(writer, index=False)
    else:
        # Load the existing workbook and add a new sheet
        book = load_workbook(excel_path)
        with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
            writer.book = book
            writer.sheets = {ws.title: ws for ws in book.worksheets}
            df.to_excel(writer, index=False, sheet_name="Pin Configurations")
            writer.save()

# File paths and terms
file_path = 'G:\\My Drive\\Study\\UQS\\ELE2303 - Embedded Systems Design\\chips\\PIC18F2525.PDF'
search_terms = ['TRISB', 'TRISA', 'PORTB', 'PORTA', 'LATB', 'LATA', 'ANSEL']
excel_path = 'G:\\My Drive\\Study\\UQS\\ELE2303 - Embedded Systems Design\\Assesments\\3\\Book2.xlsx'

# Execute the functions
results = search_pdf(file_path, search_terms)
export_to_excel(results, excel_path)
```
