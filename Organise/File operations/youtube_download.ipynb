```python
from pytube import YouTube

def download_song(url, output_path='songs'):
    yt = YouTube(url)
    audio_stream = yt.streams.filter(only_audio=True).first()
    audio_stream.download(output_path=output_path)

# Example usage
download_song('https://www.youtube.com/watch?v=oqbGus78Yig&ab_channel=MY-Topic')
```

```python
import yt_dlp

def download_audio(url, output_path='songs'):
    ydl_opts = {
        'format': 'bestaudio/best',
        'extractaudio': True,
        'audioformat': 'mp3',
        'outtmpl': f'{output_path}/%(title)s.%(ext)s',  # Save files with the video title
        'noplaylist': True,  # Avoid downloading playlists
    }

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

# Example usage
download_audio('https://www.youtube.com/watch?v=6uP-b6IiXY0&ab_channel=DANIELSHAKE')
```
