remove duplicates

```python
import os
import shutil

def get_file_paths(directory):
    """Get a list of file names in a directory."""
    file_names = set()
    for _, _, files in os.walk(directory):
        for file in files:
            file_names.add(file)
    return file_names

def delete_common_files(folder1, folder2):
    """Delete files from folder1 that are present in folder2."""
    folder1_files = get_file_paths(folder1)
    folder2_files = get_file_paths(folder2)

    common_files = folder1_files.intersection(folder2_files)

    for file in common_files:
        file_path = os.path.join(folder1, file)
        if os.path.exists(file_path):
            os.remove(file_path)
            print(f"Deleted: {file}")
 
# Specify the paths to the two folders
folder1_path = input("Enter the path to Folder 1: ")
folder2_path = input("Enter the path to Folder 2: ")

delete_common_files(folder1_path, folder2_path)
```

Delete all desktop.ini files in a directory tree:

```python
import os

def delete_desktop_ini(root_directory):
    for dirpath, _, filenames in os.walk(root_directory):
        for filename in filenames:
            if filename.lower() == 'desktop.ini':
                file_path = os.path.join(dirpath, filename)
                try:
                    os.remove(file_path)
                    print(f"Deleted: {file_path}")
                except Exception as e:
                    print(f"Error deleting {file_path}: {e}")

# Replace 'your_directory_path' with the path to your target directory
delete_desktop_ini('your_directory_path')
```

```python
import os

# Directory containing the files
directory = "H:/My Drive/Books"
# Text to remove
text_to_remove = "[rutracker-2726399]"

# Loop through the files in the directory
for filename in os.listdir(directory):
    # Full file path
    old_file = os.path.join(directory, filename)
    # Check if it is a file
    if os.path.isfile(old_file):
        # Remove the specified text from the filename
        new_filename = filename.replace(text_to_remove, "")
        # Full new file path
        new_file = os.path.join(directory, new_filename)
        # Rename the file
        os.rename(old_file, new_file)
```

convert from .webp to .jpeg

```python
from PIL import Image
import os

# Path to the directory containing the images
input_dir = 'C:/Users/ilo_l/Downloads'
output_dir = 'G:/My Drive/Code/Tarot/img/Mystical Dreams'

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Function to convert .webp to .jpeg
def convert_webp_to_jpeg(input_dir, output_dir):
    for file_name in os.listdir(input_dir):
        if file_name.endswith('.webp'):
            image_path = os.path.join(input_dir, file_name)
            image = Image.open(image_path)
            new_file_name = os.path.splitext(file_name)[0] + '.jpeg'
            output_path = os.path.join(output_dir, new_file_name)
            image.convert("RGB").save(output_path, "JPEG")
            print(f"Converted {file_name} to {new_file_name}")

convert_webp_to_jpeg(input_dir, output_dir)

output_dir  # Show the path to the converted files
```

```python
from PIL import Image
import os
import shutil

# Define source and destination directories
source_dir = "G:/My Drive/Code/Tarot/divina-fortuna-tarot"  # Change to your source directory
destination_dir = "C:/Users/ilo_l/Downloads/wordpress-ready-images"  # Change to your destination directory

# Create destination directory if it doesn't exist
if not os.path.exists(destination_dir):
    os.makedirs(destination_dir)

# Define the compression quality (0 to 100)
compression_quality = 85

# Function to rename and compress images
def process_images(source_dir, destination_dir, compression_quality):
    for root, dirs, files in os.walk(source_dir):
        for file in files:
            if file.endswith(('.png', '.jpg', '.jpeg')):
                # Original file path
                source_file_path = os.path.join(root, file)
                
                # Create the new file name
                new_file_name = file.replace(" ", "_").replace("-", "_")  # Replace spaces and hyphens with underscores
                
                # Create destination file path
                destination_file_path = os.path.join(destination_dir, new_file_name)
                
                # Open the image
                img = Image.open(source_file_path)
                
                # Save the image with compression
                img.save(destination_file_path, optimize=True, quality=compression_quality)
                
                print(f"Processed {source_file_path} -> {destination_file_path}")

# Run the function
process_images(source_dir, destination_dir, compression_quality)
```
