```python
import qrcode

# Input URL
url = "https://forms.gle/XrTEBAcz7uX2eHLBA"  # Replace with your desired URL

# Generate the QR Code
qr = qrcode.QRCode(
    version=1,  # Version controls the size of the QR code (1 is smallest)
    error_correction=qrcode.constants.ERROR_CORRECT_L,  # Error correction level
    box_size=10,  # Size of each box in the QR code grid
    border=4,  # Border size (minimum is 4 for most QR readers)
)

# Add the URL to the QR code
qr.add_data(url)
qr.make(fit=True)

# Create an image of the QR Code
img = qr.make_image(fill_color="black", back_color="white")

# Save the QR Code as an image
img.save("qrcode.png")  # Saves the QR code as "qrcode.png"
```
